```python
import sqlite3
conn=sqlite3.connect('test.db')
cursor=conn.cursor()
```

```python
cursor.execute("""
create table students(
stud_id integer primary key,
stu_name text,
class text,
marks integer)""")
conn.commit()
```

```python
cursor.executemany("""insert into students(stud_id,stu_name,class,marks)values(?,?,?,?)""",[
    (1,'priya','10A',87),
    (2,'Divya','10A',78),
    (3,'anu','10B',90),
    (4,'praga','10B',98),
    (5,'viji','10C',89)
])
conn.commit()
```

```python
cursor.execute("""select count(*) as total_students,
avg(marks) as avg_students,
max(marks) as highest_students,
min(marks) as lowest_students
from students""")
for row in cursor.fetchone():
  print(row)
```

```python
import sqlite3
conn=sqlite3.connect("hospital_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table hos(
patient_id integer primary key,
age integer,
city text,
disease text,billamount integer)""")
conn.commit()
```

```python
patient_data=[
    (1,21,'chennai','fever',500),
    (2,23,'tanjore','cold',200),
    (3,35,'coimbatore','viral fever',700),
    (4,25,'chennai','leg pain',1000),
    (5,34,'trichy','knee pain',15000)
    ]
cursor.executemany("insert into hos values(?,?,?,?,?)",patient_data)
conn.commit()

```

```python
cursor.execute("select patient_id,age,city from hos where age between 20 and 30 and disease not in('cold','fever')")
for row in cursor.fetchall():
  print(row)
```

```python
cursor.execute("select distinct disease from hos")
for row in cursor.fetchall():
  print(row)
```

```python
cursor.execute("select patient_id,age from hos where city='trichy'or 'tanjore'")
for row in cursor.fetchall():
  print(row)
```

```python
import sqlite3
conn=sqlite3.connect("product_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table prot(
produ_id integer primary key,
prod_name text,
category text,
price integer,
stock integer)""")
conn.commit()
```

```python
prod_data=[
    (1,'laptop','electronis',30000,5),
    (2,'tablet','electronics',20000,6),
    (3,'chair','furniture',3000,4),
    (4,'fan','appliances',3000,7),
    (5,'mobile','electronics',30000,8)

]
cursor.executemany("insert into prot values(?,?,?,?,?)",prod_data)
conn.commit()
```

```python
cursor.execute("select produ_id,prod_name,category from prot where price between 15000 and 20000")
for row in cursor.fetchall():
  print(row)
```

```python
cursor.execute("select produ_id,prod_name,category from prot where category in ('electronics','appliances')")
for row in cursor.fetchall():
  print(row)
```

```python
cursor.execute("select produ_id,prod_name,category from prot where stock not between 0 and 5")
for row in cursor.fetchall():
  print(row)
```

```python
cursor.execute("select distinct category from prot")
for row in cursor.fetchall():
  print(row)
```

```python
cursor.execute("select price, prod_name as item_name from prot order by price desc limit 3")
for row in cursor.fetchall():
  print(row)
```
