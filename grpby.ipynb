```python
import sqlite3
conn=sqlite3.connect("memory")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table employees(
emp_id integer primary key,
name text,
salary integer)""")
```

```python
cursor.execute("alter table employees add column department text")
```

```python
cursor.executemany("insert into employees values(?,?,?,?)",[
    (1,'asha',50000,'IT'),
    (2,'bala',4000,'HR'),
    (3,'cjith',6000,'IT'),
    (4,'deepak',None,'finance')
])
```

```python
cursor.execute("delete from employees where salary is Null or salary<0")
```

```python
cursor.execute("select * from employees")
for i in cursor.fetchall():
  print(i)
```

```python
cursor.execute("""
create table staf(
emp_name text,
department text,
salary integer)""")
```

```python
cursor.executemany("insert into  staf values(?,?,?)",[
    ('asha','IT',70000),
    ('bala','HR',70000),
    ('cijitu','manager',75000),
    ('anita','IT',80000),
    ('karthik','HR',90000)])
```

```python
cursor.execute("""
select emp_name,department,salary, rank() over (order by salary desc) as rank from staf""")
for row in cursor.fetchall():
  print(row)
```

```python
cursor.execute("""
SELECT emp_name, department, salary,
       RANK() OVER (ORDER BY salary asc) AS rank
FROM staf
""")

for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute("""
select emp_name,department,salary,dense_rank() over(order by salary asc) as dense_rank from staf""")
for row in cursor.fetchall():
  print(row)
```

```python
cursor.execute("""
SELECT department, COUNT(*) AS employee_count
FROM staf
GROUP BY department
""")

for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute("""
SELECT salary, COUNT(*)
FROM staf
GROUP BY salary
HAVING COUNT(*) >= 2
""")

for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute("""
select salary,count(*)
from staf
group by salary
having count(*)>1
""")
for row in cursor.fetchall():
  print(row)
```

```python
cursor.execute("""
update employees set salary=case
when department='IT' then salary*1.10
when department='HR'then salary*1.05
else salary end""")
for row in cursor.execute("select * from employees"):
  print(row)
```

```python
cursor.execute("""
select emp_name,salary
from staf
where salary>=(select max(salary) from staf where department='HR')""")
for row in cursor.fetchall():
  print(row)
```
