```python
import sqlite3
conn=sqlite3.connect("college_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table employee(
emp_id integer primary key,
emp_name text not null,
age integer check( age>=18),
email text unique,
course text default'HR',
fees_paid real check(fees_paid>0)
)""")
```

```python
cursor.executemany("insert into employee(emp_id,emp_name,age,email,fees_paid)values(?,?,?,?,?)",[
                   (1,'priya',18,'priya@gmail.com',25000),
                   (2,'anu',20,'anu@gmail.com',30000),
])
```

```python
import sqlite3
conn=sqlite3.connect("memory_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
    create table employee(
    emp_id integer primary key autoincrement,
    emp_name text not null,
    department taxt check(department in('hr','IT','finance','marketing')),
    salary real check(salary>15000),
    phone text unique,
    joining_date text not null
    )
""")
```

```python
cursor.executemany("""insert into employee(emp_name,department,salary,phone,joining_date)values('suresh','IT',23000,9876543209,2024)""")
```
