```python
import pandas as pd
df=pd.read_excel("products_data.xlsx")
p=df[(df["Category"]=='Toys')|(df["Category"]=="sports")]
print(p)
print()
t=df[-(df["Stock"]>0)]
print(t)
print()
d=df.loc[(df["Price"]>=1000)&(df["Price"]<=5000),["Product Name","Price"]]
print(d)
print()
o=df.iloc[0:20]
print(o[o["Category"]=="Electronics"])
print()
p=["USA","UK","Canada"]
result=df[df["Country"].isin(p)]
print(result)
print()
e=["Books","Beauty"]
y=df[-df["Category"].isin(e)]
print(y)
print()
print(df["Country"].nunique())
print()
result=df[df["Rating"].between(4.0,5.0)]
print(result)
print()
p=df[df["Product Name"].str.contains("wireless",case=False)]
print(p)
print()
c=["Product Name","Category","Price","Rating","Stock","Country"]
d=df[df.duplicated(subset=c,keep="first")]
print(d)
print()
t=df.nlargest(5,"Price")
b=df.nsmallest(5,"Rating")
print(t)
print(b)
```
