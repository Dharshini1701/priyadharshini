```python
import sqlite3
conn=sqlite3.connect("test_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table students(
stu_id integer primary key,
stu_name text,
stu_age integer)
""")
conn.commit()
```

```python
student_data=[
    (1,'priya',18),
    (2,'Anu',19),
    (3,'raju',20)
]
cursor.executemany("insert into students values(?,?,?)",student_data)
conn.commit()
```

```python
cursor.execute("""
create table deparment(
dep_id integer primary key,
dep_name text
)""")
```

```python
studen_data=[
    (1,"anu"),
    (2,'priya'),
    (3,'raju')
]
cursor.executemany("insert into deparment values(?,?)",studen_data)
conn.commit()
```

```python

cursor.execute("""
select
s.stu_id,
s.stu_name,
d.dep_id
from students s inner join deparment d on s.stu_id=d.dep_id""")
print(cursor.fetchall())



```

```python
import sqlite3
conn=sqlite3.connect("employee_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table emplo(
employee_id integer primary key,
employee_name text,
department text,
salary integer)""")
conn.commit()
```

```python
employee_data=[
    (12,'priya','IT',80000),
    (13,'anu','Manager',70000),
    (14,'raju','HR',80000),
    (15,'nithi','IT',70000),
    (16,'sakhtu','bpo',60000)
]
cursor.executemany("insert into emplo values(?,?,?,?)",employee_data)
conn.commit()
```

```python
cursor.execute("""
create table dep(
dep_id integer primary key,
dep_name text,
dep_location text)""")
conn.commit()
```

```python
dept_data=[
    (1,'IT','chennai'),
    (2,'Manager','trichy'),
    (3,'HR','chennai'),
    (4,'IT','trichy'),
    (5,'bpo','trichy')
]
cursor.executemany("insert into dep values(?,?,?)",dept_data)
conn.commit()
```

```python
cursor.execute("""
select
e.employee_name,
e.department,
e.salary,
d.dep_location from emplo e cross join dep d""")
for row in cursor.fetchall():
  print(row)
```

```python
import sqlite3
conn=sqlite3.connect("employee_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table emp(
emp_id integer primary key,
emp_name text,
emp_age integer
)""")
conn.commit()
```

```python
emp_data=[
    (1,'priya',21),
    (2,'Anu',22),
    (3,'raju',23)
]
cursor.executemany("insert into emp values(?,?,?)",emp_data)
conn.commit()
```

```python
cursor.execute("""
create table manager(
manager_id integer primary key,
manager_name text,
department text)""")
conn.commit()
```

```python
man_data=[
    (1,'nithiya','CSE'),
    (2,'priya','CS'),
    (3,'raji','CSE')
]
cursor.executemany("insert into manager values(?,?,?)",man_data)
conn.commit()
```

```python
cursor.execute("""
select emp_id,emp_name from emp union
select manager_id,manager_name from manager""")
for row in cursor.fetchall():
  print(row)
```
