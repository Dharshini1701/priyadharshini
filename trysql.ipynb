```python
import sqlite3
conn=sqlite3.connect("college_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table college(
student_id integer primary key,
name text,
department text,
marks integer)
""")
conn.commit()
```

```python
college_data=[
    (1,'arun','CSE',85),
    (2,'divya','ECE',90),
    (3,'Karthick','IT',78),
    (4,'meena','CSE',88),
    (5,'rahul','EEE',72)
]
cursor.executemany("insert into college values(?,?,?,?)",college_data)
conn.commit()
```

```python
cursor.execute("select* from college where department='CSE'")
for row in cursor.fetchall():
  print(row)
conn.commit()
```

```python
cursor.execute("select * from college where department='CSE' and marks>85")
for row in cursor.fetchall():
  print(row)
conn.commit()
```

```python
cursor.execute("select * from college where student_id=6 or name='divya'")
for row in cursor.fetchall():
  print(row)
conn.commit()
```

```python
cursor.execute("select * from college where not department= 'ECE'")
for row in cursor.fetchall():
  print(row)
conn.commit()
```

```python
cursor.execute("select * from college where department in('CSE','ECE')")
for row in cursor.fetchall():
  print(row)
conn.commit()
```

```python
cursor.execute("select * from college order by marks asc")
for i in cursor.fetchall():
  print(i)
conn.commit()
```

```python
cursor.execute("select * from college order by marks desc")
for i in cursor.fetchall():
  print(i)
conn.commit()
```

```python
cursor.execute("select count(*) from college")
print(cursor.fetchone()[0])
```

```python
cursor.execute("select sum(marks),avg(marks) from college")
print(cursor.fetchone())
```
