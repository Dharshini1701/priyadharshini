```python
import sqlite3
conn=sqlite3.connect("student_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table course(
course_id integer primary key,
course_name text,
duration_months integer

)""")
conn.commit()
```

```python
cursor.execute("""
create table stud(
stu_id integer primary key,
stu_name text,
age integer,
course_id integer,
foreign key(course_id) references course(course_id))""")
conn.commit()
```

```python
cour=[
    (1,'cs',3),
    (2,'ds',2),
    (3,'AI',6)

]
cursor.executemany("insert into course values(?,?,?)",cour)
conn.commit()
```

```python
cursor.execute("select s.stu_id,s.stu_name,c.course_name,s.age from stud s join course c on s.course_id=c.course_id")
for row in cursor.fetchall():
  print(row)
```

```python
import sqlite3
conn=sqlite3.connect("books_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table book(
    book_id integer primary key,
    title text,
    author text
)""")
conn.commit()
```

```python
cursor.execute("""
create table member(
member_id integer primary key,
name text,
email text)
""")
conn.commit()
```

```python
cursor.execute("""
create table loan(
loans_id integer primary key,
book_id integer,
member_id integer,
loan_date integer,
foreign key(book_id) references book(book_id),
foreign key(member_id) references member(member_id))""")
conn.commit()
```

```python
book=[
    (1,'java','guido'),
    (2,'python','monty'),
    (3,'c','rossum'),
    (4,'c++','bjante')
]
cursor.executemany("insert into book values(?,?,?)",book)
conn.commit()
```

```python
import sqlite3
conn=sqlite3.connect("customers_db")
cursor=conn.cursor()
```

```python
cursor.execute("""
create table custom(
cust_id integer primary key,
name text,
city text
)""")
conn.commit()
```

```python
cursor.execute("""
create table ordss(
order_id integer primary key,
cust_id integer,
order_date integer,
foreign key (cust_id)references custom(cust_id))""")
conn.commit()
```

```python
cursor.execute("""
create table ord_item(
ord_id integer primary key,
order_id integer,
product_name text,
quantity integer,
price integer,
foreign key(order_id)references ordss(order_id))""")
conn.commit()
```

```python
cus=[
    (1,'priya','chennai'),
    (2,'anu','trichy'),
    (3,'viji','madurai')
]
cursor.executemany("insert into custom values(?,?,?)",cus)
conn.commit()
```

```python
odd=[
    (11,1,2022),
    (12,4,2021),
    (13,3,2020),
    (14,2,2019)
]
cursor.executemany("insert into ordss values(?,?,?)",odd)
conn.commit()
```

```python
ods=[
    (111,2,'laptop',5,20000),
    (112,4,'mouse',6,30000),
    (113,5,'monitor',5,25000),
    (114,3,'keyboard',4,20000)

]
cursor.executemany("insert into ord_item values(?,?,?,?,?)",ods)
conn.commit()
```

```python
cursor.execute("""
SELECT c.name,
       SUM(oi.quantity * oi.price) AS total_amount
FROM custom c
JOIN ordss o
    ON c.cust_id = o.cust_id
JOIN ord_item oi
    ON o.order_id = oi.order_id
GROUP BY c.name
""")
for row in cursor.fetchall():
  print(row)
```
